# Data Collection

## Project: AI-Assisted Pull Request Outcome Prediction

In this step, we collect and organize the real-world data required for our machine learning project.

Our project uses AI-assisted Pull Request data collected from GitHub through the AIDev dataset.

The required AIDev data is collected in Parquet format and includes Pull Request, repository, and user information.

The Pull Request dataset contains a large number of records, so Parquet format allows us to store and work with the data efficiently.

```python
# Import libraries required for working with our Parquet dataset

import pyarrow.parquet as pq
import pandas as pd
```

```python
# Locate and open the original Pull Request dataset

file_path = "../data/raw/all_pull_request.parquet"

file = pq.ParquetFile(file_path)
```

```python
# Check the total rows, row groups, and columns in the dataset

print("Number of rows:", file.metadata.num_rows)
print("Number of row groups:", file.num_row_groups)
print("Number of columns:", len(file.schema.names))
```

```python
# Display all columns available in the Pull Request dataset

print("Columns:")
for column in file.schema.names:
    print(column)
```

```python
# Read a small sample instead of loading the entire 1.5 GB dataset

sample = file.read_row_group(0).to_pandas()
sample = sample.head(10000)

print("Sample shape:", sample.shape)
```

```python
# View the first few Pull Request records

sample.head()
```

```python
# Check the sample's columns, data types, and non-null values

sample.info()
```

---

```python
# Locate and open the original AIDev repository metadata dataset.

repository_path = "../data/raw/all_repository.parquet"

repository_file = pq.ParquetFile(repository_path)
```

```python
# Check the total rows, row groups, and columns in the repository dataset.

print("Number of rows:", repository_file.metadata.num_rows)
print("Number of row groups:", repository_file.num_row_groups)
print("Number of columns:", len(repository_file.schema.names))
```

```python
# Display all columns available in the repository dataset.

print("Repository columns:")
for column in repository_file.schema.names:
    print(column)
```

```python
# Read a small repository sample for inspection.

repository_sample = repository_file.read_row_group(0).to_pandas()
repository_sample = repository_sample.head(10000)

print("Repository sample shape:", repository_sample.shape)
```

```python
# View the first few repository records.

repository_sample.head()
```

```python
# Check the repository sample's columns, data types, and non-null values.

repository_sample.info()
```

---

```python
# Locate and open the original AIDev user metadata dataset.

user_path = "../data/raw/all_user.parquet"

user_file = pq.ParquetFile(user_path)
```

```python
# Check the total rows, row groups, and columns in the user dataset.

print("Number of rows:", user_file.metadata.num_rows)
print("Number of row groups:", user_file.num_row_groups)
print("Number of columns:", len(user_file.schema.names))
```

```python
# Display all columns available in the user dataset.

print("User columns:")
for column in user_file.schema.names:
    print(column)
```

```python
# Read a small user sample for inspection.

user_sample = user_file.read_row_group(0).to_pandas()
user_sample = user_sample.head(10000)

print("User sample shape:", user_sample.shape)
```

```python
# View the first few user records.

user_sample.head()
```

```python
# Check the user sample's columns, data types, and non-null values.

user_sample.info()
```

# Saving Sample Data

```python
# Save inspection samples from the Pull Request, Repository, and User datasets as Excel files.

sample.to_excel(
    "../data/sample/pr_sample.xlsx",
    index=False
)

repository_sample.to_excel(
    "../data/sample/repository_sample.xlsx",
    index=False
)

user_sample.to_excel(
    "../data/sample/user_sample.xlsx",
    index=False
)

print("PR sample saved successfully.")
print("Repository sample saved successfully.")
print("User sample saved successfully.")
```
